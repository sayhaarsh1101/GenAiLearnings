```python
from langchain_community.embeddings import OllamaEmbeddings
from langchain_ollama import ChatOllama
```

```python
embeddings = OllamaEmbeddings(
    model="nomic-embed-text"
)
```

```python
embeddings
```

```python
r1=embeddings.embed_documents(
    [
       "Alpha is the first letter of Greek alphabet",
       "Beta is the second letter of Greek alphabet", 
    ]
)
```

```python
len(r1[0])
```

```python
r1[1]
```

```python
embeddings.embed_query("What is the second letter of Greek alphabet ")
```

```python
llm = ChatOllama(
    model="llama3", # or "phi4-mini"
    temperature=0.3,  # Lower for more factual answers
)
```

```python
# Example Usage:
question = "What was the salt march?"
# Step 1: Embeddings help find the right Wikipedia docs (we'll do this next)
# Step 2: LLM answers the question
response = llm.invoke(f"Based on the context provided, answer this: {question}")
print(response.content)
```
